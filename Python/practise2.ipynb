```python
def fact(x):
    total = 1
    while x>0:
        total = total * x
        x = x-1
    return total
```

```python
fact(3)
```

```python
def fact1(x):
    if x == 0 or x == 1:
        return 1
    return x * fact1(x - 1)

```

```python
fact1(3)
```

```python
def palindrome(x):
    y=[]
    for i in x:
        y.append(i)
    if x == "".join(y[::-1]):
        print("its palindrome")
    else:
        print("no")
        return

```

```python
x = "momos"
palindrome(x)
```

```python
def num_palin(a):
    y = a
    temp = 0
    while y>0:
        z =y%10
        temp = temp*10 + z
        y = y//10
    if a == temp:
        print("yes")
    else:
        print("no")
    return
```

```python
num_palin(121)
```

```python
def sum_digits(x):
    c = x
    v = 0
    for i in str(c):
        v = v+int(i)
    return v
```

```python
def vowel_count(x):
    count = 0
    vo = ['a','e','i','o','u']

    for i in x:
        if i in vo:
            count += 1
        else:
            continue
    return count
```

```python
x = "aeiouaeiou"
vowel_count(x)
```

```python
def prime(x):
    if x <=1:
        print(f"{x} is a prime number")
    else:
        is_prime = True
        for i in range(2,int(x**0.5)+1):
            if x%i==0:
                is_prime = False
                break
        if is_prime:
            print("True")
        else:
            print("False")
        return


    
```

```python
x = 20
prime(x)
```

```python
def max_digit(x):
    temp = 0
    for i in str(x):
        if int(i)>temp:
            temp = int(i)
        
    return temp    
    
```

```python
x = 1234567895
max_digit(x)
```

```python
def calculator(a,b,op):
    if op == '+':
        return(a+b)
    elif op == '-':
        return(a-b)
    elif op == '*':
        return(a*b)
    elif op == '%':
        return(a%b)
    elif op == '**':
        return(a**b)

```

```python
calculator(2,3,'+')
```
