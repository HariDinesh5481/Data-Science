```python
def chec(*numbers):
    max_num =numbers[0]
    min_num =numbers[0]
    for i in numbers:
        if i > max_num:
            max_num=i
        elif i < min_num:
            min_num=i
    return(max_num,min_num)
```

```python
T = int(input())

for _ in range(T):
    N = int(input())
    arr = list(map(int, input().split()))
    
    min_index = arr.index(min(arr))
    max_index = arr.index(max(arr))
    
    swaps = min_index + (N - 1 - max_index)
    
    if max_index < min_index:
        swaps -= 1
    
    print(swaps)
```

```python
def solve():
    T = int(input())
    
    for _ in range(T):
        N = int(input())
        S = list(map(int, input().split()))
        K = int(input())

        S.sort(reverse=True)  # largest first
        
        total = 0
        for i in range(K):
            total += (S[i] - 1)

        print(total)

if __name__ == "__main__":
    solve()
```

```python
for i in range(1,21):
    if i<=1:
        print("its a prime number")
    else:
        is_prime = True
        for j in range(2,int(i**0.5)+1):
            if i%j==0:
                is_prime=False
                break
        if is_prime :
            print(f"{i} is a prime ")
        else:
            print(f"{i} is not a prime number")
```
