```python
T = int(input())

for _ in range(T):
    N = int(input())
    arr = list(map(int, input().split()))
    
    min_index = arr.index(min(arr))
    max_index = arr.index(max(arr))
    
    swaps = min_index + (N - 1 - max_index)
    
    if max_index < min_index:
        swaps -= 1
    
    print(swaps)
```

```python
def solve():
    T = int(input())
    
    for _ in range(T):
        N = int(input())
        S = list(map(int, input().split()))
        K = int(input())

        S.sort(reverse=True)  # largest first
        
        total = 0
        for i in range(K):
            total += (S[i] - 1)

        print(total)

if __name__ == "__main__":
    solve()
```

```python
def is_armstrong(num):
    temp = num
    n=len(str(num))
    total_sum = 0

    while temp >0:
        digit = temp %10
        total_sum += digit ** n
        temp = temp //10

    return num == total_sum
is_armstrong(153)
```
